### Task 06 - Solution

```python
# ===== TASK 06 =====
adminUnits = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
pop17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
pop23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def toDict(turnPairs):
    return {str(ki): (value.item() if hasattr(value, "item") else value) for ki, value in turnPairs}

popTotal = pop23.sum()
print(f"T6a >> total population 2023 = {popTotal:.2f} million")

percentShare = 100 * pop23 / popTotal
print(f"T6b >> share of total (%): {toDict(zip(adminUnits, np.round(percentShare, 2)))}")

print(f"T6c >> most populous unit: {adminUnits[np.argmax(pop23)]}")
print(f"T6d >> units above 20 million: {adminUnits[np.where(pop23 > 20)]}")

yearlyGrowth = (np.power(pop23 / pop17, 1 / 6) - 1) * 100
print(f"T6e >> annual growth 2017-2023 (% per year): {toDict(zip(adminUnits, np.round(yearlyGrowth, 2)))}")

print(f"T6f >> fastest-growing unit: {adminUnits[np.argmax(yearlyGrowth)]} ({yearlyGrowth.max():.2f}% per year)")

nat = ((pop23.sum() / pop17.sum()) ** (1 / 6) - 1) * 100
doublingTime = 70 / yearlyGrowth.max()
print(f"T6 interpret >> national growth = {nat:.2f}% per year.")
print(f"T6 interpret >> {adminUnits[np.argmax(yearlyGrowth)]} would double in about {doublingTime:.0f} years at its current rate.")
print(f"T6 interpret >> AJK grew {yearlyGrowth[-1]:.2f}% per year vs national {nat:.2f}% -> "
      f"AJK is {nat - yearlyGrowth[-1]:.2f} points below the national rate (in this sample AJK is basically flat).")
```

### Task 07 - Solution

```python
# ===== TASK 07 =====
cityArr = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
monthList = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

tempGrid = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rain = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def toDict(turnPairs):
    return {str(ki): (value.item() if hasattr(value, "item") else value) for ki, value in turnPairs}

print(f"T7a >> shape of temp (cities, months): {tempGrid.shape}")

annualMean = tempGrid.mean(axis=1)
print(f"T7b >> yearly mean temperature (deg C): {toDict(zip(cityArr, np.round(annualMean, 1)))}")

peakMonth = monthList[tempGrid.argmax(axis=1)]
print(f"T7c >> hottest month per city: {toDict(zip(cityArr, peakMonth))}")

print(f"T7d >> 5-city monthly average (deg C): {toDict(zip(monthList, np.round(tempGrid.mean(axis=0), 1)))}")

nAbove35 = np.count_nonzero(tempGrid > 35, axis=1)
print(f"T7e >> months above 35 deg C: {toDict(zip(cityArr, nAbove35))}")

fLahore = tempGrid[0] * 1.8 + 32
print(f"T7f >> Lahore monthly temperature (deg F): {np.round(fLahore, 1)}")

monsoon = rain[:, 6:9].sum(1)
print(f"T7g >> monsoon rain Jul-Sep (mm): {toDict(zip(cityArr, monsoon))}")
print(f"T7g >> wettest monsoon city: {cityArr[np.argmax(monsoon)]} ({monsoon.max()} mm)")

percentShare = 100 * monsoon / rain.sum(axis=1)
print(f"T7h >> monsoon share of annual rain (%): {toDict(zip(cityArr, np.round(percentShare, 1)))}")

ki = np.where(cityArr == "Karachi")[0][0]
print(f"T7 interpret >> Karachi has the highest yearly mean ({annualMean.max():.1f} deg C) "
      f"but {nAbove35[ki]} months above 35 deg C, while cooler-on-average Peshawar has {nAbove35[4]} "
      f"and Lahore {nAbove35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

### Task 08 - Solution

```python
# ===== TASK 08 =====
yearsFy = np.array(["FY2023", "FY2024", "FY2025"])
monthFy = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

infl = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def toDict(turnPairs):
    return {str(ki): (value.item() if hasattr(value, "item") else value) for ki, value in turnPairs}

avgInfl = infl.mean(axis=1)
print(f"T8a >> average inflation per year (%): {toDict(zip(yearsFy, np.round(avgInfl, 2)))}")

fy24Vals = infl[1]
print(f"T8b >> FY2024 months above 25 %: {monthFy[fy24Vals > 25]} -> {np.sum(fy24Vals > 25)} months")

diffArr = infl[:, 1:] - infl[:, :-1]
print(f"T8c >> month-on-month change (percentage points; negative = drop):\n{np.round(diffArr, 1)}")

bigDrop = np.argmin(diffArr, axis=1)
print(f"T8d >> month of biggest drop: {toDict(zip(yearsFy, monthFy[bigDrop + 1]))} | size (pp): {np.round(diffArr.min(axis=1), 1)}")

fy25 = infl[2]
labels = np.where(fy25 < 5, "Low", np.where(fy25 < 10, "Moderate", "High"))
print(f"T8e >> FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((monthFy, fy25, labels))}")

print(f"T8 interpret >> average inflation went {avgInfl[0]:.1f}% -> {avgInfl[1]:.1f}% -> {avgInfl[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

### Task 09 - Solution

```python
# ===== TASK 09 =====
examMarks = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
rollIds = np.arange(1, 21)
THRESHOLD = 40

newMarks = np.clip(examMarks + 5, 0, 100)
print(f"T9a >> original marks: {examMarks}")
print(f"T9a >> marks after +5 grace (cap 100): {newMarks}")

passBefore = np.mean(examMarks >= THRESHOLD) * 100
pctAfter = 100 * np.mean(newMarks >= THRESHOLD)
print(f"T9b >> pass % before grace = {passBefore:.1f} %, after grace = {pctAfter:.1f} %")

graceMask = np.logical_and(examMarks < THRESHOLD, newMarks >= THRESHOLD)
print(f"T9c >> passed only because of grace (roll numbers): {rollIds[graceMask]}")
print(f"T9c >> their original marks: {examMarks[graceMask]}")

zs = (newMarks - newMarks.mean()) / newMarks.std()
print(f"T9d >> mean = {newMarks.mean():.2f}, std = {newMarks.std():.2f}")
print(f"T9d >> z-scores (std devs from mean): {np.round(zs, 2)}")
```

### Task 10 - Solution

```python
# ===== TASK 10 =====
canvas = np.zeros((8, 8), dtype=np.uint8)
canvas[1, 1:7] = 255
for r, c in zip([2, 3, 4, 5, 6], [6, 5, 5, 4, 4]):
    canvas[r, c] = 255

render = lambda a: (a > 0).astype(int)

print(f"T10a >> the '7' (1 = white):\n{render(canvas)}")

imgFlip = canvas[:, ::-1]
print(f"T10b >> flipped left-right:\n{render(imgFlip)}")

invImg = 255 - canvas
print(f"T10c >> inverted (white <-> black):\n{render(invImg)}")

subImg = canvas[:7, 1:]
print(f"T10d >> cropped 7x7 region, shape: {subImg.shape}\n{render(subImg)}")

bright = np.clip(canvas.astype(np.int16) + 60, 0, 255).astype(np.uint8)
print(f"T10e >> brightened (+60, capped at 255); unique values: {np.unique(bright)}")

print(f"T10f >> shape = {canvas.shape} | dtype = {canvas.dtype} | white pixels = {(canvas == 255).sum()}")
```
